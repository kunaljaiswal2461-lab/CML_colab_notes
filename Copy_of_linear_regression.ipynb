```python
import pandas as pd
from sklearn.datasets import fetch_california_housing
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = fetch_california_housing()
df
```

```python
data = pd.DataFrame(data = df.data , columns=df.feature_names)
data['MedHouseVal'] = df.target
display(data.head())
```

```python
#independent and dependent features
y = df.target
X = data.drop('MedHouseVal', axis=1)
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42,
)

```

```python
X_train
```

```python
for i in y_train:
  print(i)
```

```python
#standardize the input data
from sklearn.preprocessing import StandardScaler
Scaler = StandardScaler()
```

```python
X_train = Scaler.fit_transform(X_train)
```

```python
X_train
```

```python
X_test = Scaler.transform(X_test)
X_test
```

```python
from  sklearn.linear_model import LinearRegression
```

```python
from sklearn.model_selection import cross_val_score
regression = LinearRegression()
regression.fit(X_train, y_train)
```

```python
mse = cross_val_score(regression,X_train,y_train,scoring = "neg_mean_squared_error" , cv = 1000)
```

```python
mse
```

```python
mean = np.mean(mse)
mean
```

```python
#prediction
reg_pred = regression.predict(X_test)
```

```python
for i in reg_pred:
  print(i)
```

```python
import seaborn as sns
```

```python
sns.displot(reg_pred - y_test, kind = "kde")
```

```python
from sklearn.metrics import r2_score
```

```python
score = r2_score(y_test, reg_pred)
score
```

##RIDGE REGRESSION ALGORITHM


```python
from sklearn.linear_model import Ridge
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
ridge_regressor = Ridge()
```

```python
ridge_regressor
```

```python
parameters = {'alpha':[1,2,5,10,40,30,70,100]}
ridgecv = GridSearchCV(ridge_regressor,parameters , scoring = 'neg_mean_squared_error', cv = 5)
#ridgecv.best_params_
ridgecv.fit(X_train,y_train)
```

```python
print(ridgecv.best_params_)
```

```python
print(ridgecv.best_score_)
```

```python
ridge_pred = ridgecv.predict(X_test)
```

```python
import seaborn as sns
sns.set_theme(style="whitegrid")
sns.displot(ridge_pred - y_test,kind = 'kde')
```

```python
# Correcting the argument order: r2_score(y_true, y_pred)
score = r2_score(y_test, ridge_pred)
print(f"Ridge Regression R2 Score: {score}")
```

##LASSO REGRESSION

```python
from sklearn.linear_model import Lasso
```

```python

Lasso_ = Lasso()
```

```python
parameters = {'alpha':[1,2,5,10,40,30,70,100]}
lassocv = GridSearchCV(Lasso_,parameters , scoring = 'neg_mean_squared_error', cv = 5)
#ridgecv.best_params_
lassocv.fit(X_train,y_train)
```

```python
lassocv.best_score_
```

```python
lassocv.best_params_
```

```python
pred_lasso = lassocv.predict(X_test)
```
